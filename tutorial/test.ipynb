#### Import all the module components

#### Set Parameters

This is hopefully temporary and you should only have to set some of these values by the time this code is finished. 

```python
'''Define the eight properties from: A figure-of-merit-based framework to evaluate photovoltaic materials (arxiv).'
' Use the default values from Table 2 for MAPI https://iopscience.iop.org/article/10.1088/2515-7655/ad2499/pdf'''

# Cell parameters

Tcell = 298

# Spectral/ material dependant values - could be calculated from the equations (1) and (2) in the research paper based on the spectral values (see bottom of notebook)

alpha = 67913.975760                # Spectral_average (requires materials absorption coefficient spectrum) These can now be taken from VASP absorption.dat at bottom of notebook.
sigma = 0.542762                 # sigma spectral dispersion of the absorption coefficient spectrum - These can now be taken from VASP absorption.dat at bottom of notebook.

# Material dependant values

tau = 2.5E-7                # tau non-radiative recombination life time SRH - variable (defects) - carrier capture
mu = 10                      # mu carrier mobility - AMSET - generate input and parse output files
dop_density = 1e16           # n doping density - Variable 
epsilon = 10               # epsilon static dielectric constant - Can be calculated from VASP code - wiki2
dos_mass =  0.3             # m density of states effective mass - lucydot - effmass dos.py 
E_gap = 1.12                 # E₉ Electronic Band gap - From sumo-bandstats - This probably needs to be the OPT gap?

spectrum = np.loadtxt('ASTMG173.csv', delimiter=',', skiprows=1)
```

#### Test the Figure of Merit Code

```python
photon_spectrum = DB_FOM.convert_spectrum(spectrum)

FOM = Final_results.Crovetto_efficiency(E_gap, photon_spectrum, alpha, tau, sigma, dos_mass, dop_density, epsilon, mu, Tcell)

print(FOM)
```
