# Structural Relaxation with CASTEP

The first step in most ab-initio workflows after obtaining an atomic structure and deciding on a computational regime (DFT functional, basis set, and k-point sampling) is to relax/optimise the structure.

Relaxation finds the lowest energy arrangement/equilibrium positions of the atoms in the structure with respect to the given regime. This is the structure we need to get a good description of the material's electronic properties.

Solphin provides functions to write input files and read output files to achive this using CASTEP.

1) Import the necessary libraries.

````python
import solphin.castep_inputs as castep_inputs
import solphin.vasp_inputs as vasp_inputs
````

2) Define our computational regime. These values are chosen based on previous literature, experience, or convergence testing.

````python
functional = "HSE06"      # The DFT functional
cut_off_energy = 450      # Plane-wave cut-off in eV
kpoint_mp_spacing = 0.03  # Maximum k-point spacing in Angstrom^-1, no 2*pi factor
````

3) Load the initial structure. Here we read from the bundled test data.

````python
structure = vasp_inputs.read_structure_pmg("../tests/data/Cu2GeS3/Relax/POSCAR")
````

4) Write the calculation input files. Here we write to `workdir/relax` which is untracked, and can be used to try out these tutorials.

````python
relax_out = "workdir/relax"

relax_cell = castep_inputs.write_castep_calculation(
    structure=structure,
    recipe=functional,
    out_dir=relax_out,
    patches=["relax_cell", "tight_relax"],
    kpoint_mp_spacing=kpoint_mp_spacing,
    user_param_settings={"cut_off_energy": cut_off_energy},
)
````

5. Run the calculation the way your machine wants CASTEP run — for example:

```bash
   cd workdir/relax
   mpirun -np 16 castep.mpi Cu2GeS3
```
